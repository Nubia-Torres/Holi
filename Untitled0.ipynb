```python
x= int(input("Escriba un primer número "))
y= int(input("Escriba un segundo número "))
print("La suma de los números es igual a: ",x+y)
print("La resta de los números es igual a: ",x-y)
print("La multiplicación de los números es igual a: ",x*y)
print("La división de los números es igual a: ",x/y)
```

```python
Celsius=float(input("Ingrese la temperatura en grados Celsius: "))
Fahrenheit=9/5*Celsius+32
print("La temperatura en grados Fahrenheit es: ",Fahrenheit)
```

```python
Altura = float(input("Ingrese la altura del triángulo: "))
Base = float(input("Ingrese la base del triángulo: "))
Area = (Altura * Base) / 2
print("El área del triángulo es:", Area)
```

```python
dolar=float(input("Ingrese la cantidad en dólares: "))
tasa=0.89
euro=dolar*tasa
print("La cantidad en euros es: ",euro)
```

```python
precio=float(input("Ingrese el precio del producto: "))
descuento=float(input("Ingrese el porcentaje de descuento: "))
precio_final=precio*(1-descuento/100)
print("El precio final del producto es: ",precio_final)
```
